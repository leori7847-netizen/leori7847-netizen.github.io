```python
import pandas as pd
import jieba
from sklearn.feature_extraction.text import TfidfVectorizer

data = pd.read_csv("text_classification_data.csv")
stopwords = {"的", "了", "是", "我"}
data["cleaned_content"] = data["content"].fillna("").apply(
    lambda text: " ".join(word for word in jieba.lcut(text) if word.strip() and word not in stopwords))
features = TfidfVectorizer().fit_transform(data["cleaned_content"])
print(data["cleaned_content"].head(3).to_string(index=False))
print("TF-IDF 形状：", features.shape)
data.to_csv("cleaned_text.csv", index=False)
```
