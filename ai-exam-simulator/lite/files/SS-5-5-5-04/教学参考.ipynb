```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score
data = pd.read_csv("04/SS-5-5-5-04_user_load.csv")
data = data.dropna().drop_duplicates().reset_index(drop=True)
X = data[["历史日均用电量", "历史峰值负荷", "月用电量", "用户类型", "电压等级", "温度", "湿度", "是否节假日", "时刻"]]
y = data["是否高峰"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2,
random_state=2026, stratify=y)
model = RandomForestClassifier(n_estimators=100, random_state=2026)
model.fit(X_train, y_train)
pred = model.predict(X_test)
accuracy = accuracy_score(y_test, pred)
f1 = f1_score(y_test, pred)
capability = pd.DataFrame({"指标": ["Accuracy", "F1"], "值": [accuracy,
f1]})
print(capability)
below = capability[capability["值"] < 0.80]
print("模型通过质量门禁" if len(below) == 0 else "模型需进入下一轮迭代")
```
