```python
import pandas as pd

data = pd.read_csv("business_data.csv")
data["order_date"] = pd.to_datetime(data["order_date"], errors="coerce")
data["amount"] = data["amount"].fillna(data["amount"].mean())
data = data[data["amount"] >= 0].drop_duplicates().copy()
data["amount_level"] = data["amount"].apply(lambda value: "低额" if value <= 100 else "中额" if value <= 500 else "高额")
print("总行数：", len(data))
print("均值：", data["amount"].mean())
print("最大值：", data["amount"].max())
data.to_csv("cleaned_data.csv", index=False)
```
